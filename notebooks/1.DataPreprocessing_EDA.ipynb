## Data Precprocessing

```python
import re
import emoji
import pandas as pd
import numpy as np


# Clean and normalize the text
def clean_text(text):

    # Convert the input to a string
    text = str(text)
    
    # Remove URLs
    text = re.sub(r"http\S+|www\S+", " ", text)

    # Remove user mentions
    text = re.sub(r"@\w+", " ", text)

    # Remove the '#' symbol but keep the hashtag word
    text = re.sub(r"#(\w+)", r"\1", text)

    # Remove HTML tags
    text = re.sub(r"<.*?>", " ", text)

    # Convert emojis into text
    text = emoji.demojize(text, delimiters=(" ", " "))

    # Replace underscores with spaces
    text = text.replace("_", " ")

    # Remove non-ASCII characters
    text = re.sub(r"[^\x00-\x7F]+", " ", text)

    # Remove extra spaces
    text = re.sub(r"\s+", " ", text).strip()

    # Convert text to lowercase
    return text.lower()


# Convert ratings into sentiment labels
def rating_to_sentiment(r):

    # Positive sentiment
    if r >= 4:
        return "positive"

    # Negative sentiment
    if r <= 2:
        return "negative"

    # Neutral sentiment
    return "neutral"

# Create a balanced dataset
def balanced_sample(df, per_class, seed=42):

    sampled = []

    # Group the dataset by sentiment label
    for _, group in df.groupby("sentiment_label"):

        # Randomly sample reviews from each sentiment group
        sampled.append(
            group.sample(
                n=min(len(group), per_class),
                random_state=seed
            )
        )

    # Combine all sampled groups
    return pd.concat(sampled, ignore_index=True)
    
```

```python
def standardise_amazon(df, category):

    output = pd.DataFrame({
        "platform": "amazon",
        "product_category": category,
        "text": (df["title"].fillna("") + ". " + df["text"].fillna("")).str.strip(". ").str.strip(),
        "rating": df["rating"].astype(float),
        "date": pd.to_datetime(
            df["timestamp"],
            unit="ms",
            errors="coerce"
        ).dt.strftime("%Y-%m-%d"),
        "review_id": df["asin"].astype(str) + "_" + df.index.astype(str),
    })

    # Create the sentiment label
    output["sentiment_label"] = output["rating"].apply(rating_to_sentiment)

    # Return the standardized DataFrame
    return output
```

```python
amazon_raw = {
    "All_Beauty": beauty,
    "Grocery": grocery,
    "Electronics": electronics
}

PER_CLASS = 20000
```

```python
amz = pd.concat(
    [
        balanced_sample(
            standardise_amazon(raw, cat),
            PER_CLASS
        )
        for cat, raw in amazon_raw.items()
    ],
    ignore_index=True
)

print(amz.columns.tolist())
```

### Top words per sentiment

```python
from wordcloud import WordCloud

# Define common words to exclude from the word clouds
stop = {
    "the", "a", "an", "is", "it", "was", "and", "or", "but", "to",
    "for", "of", "with", "this", "that", "not", "my", "i", "in",
    "on", "at", "you", "they", "we", "so"
}

# Create a figure with three subplots
fig, axes = plt.subplots(1, 3, figsize=(15, 5))

# Generate a word cloud for each sentiment class
for ax, s in zip(axes, order):

    # Combine all reviews for the current sentiment
    txt = " ".join(df[df["sentiment_label"] == s]["cleaned_text"])

    # Generate the word cloud
    wc = WordCloud(
        width=500,
        height=350,
        background_color="white",
        stopwords=stop
    ).generate(txt)

    # Display the word cloud
    ax.imshow(wc)
    ax.axis("off")
    ax.set_title(s.capitalize())

# Adjust the layout and save the figure
plt.tight_layout()
plt.savefig("outputs/fig_wordclouds_by_sentiment.png", dpi=150)

# Display the figure
plt.show()
```

### Reviews over time (temporal trend)

```python
# Create a copy of the dataset
tmp = df.copy()

# Extract the year and month from the review date
tmp["month"] = pd.to_datetime(
    tmp["date"],
    errors="coerce"
).dt.to_period("M").astype(str)

# Count the number of reviews by month and sentiment
trend = (
    tmp[tmp["month"] != "NaT"]
    .groupby(["month", "sentiment_label"])
    .size()
    .unstack(fill_value=0)
)

# Save the monthly sentiment trend
trend.to_csv("outputs/table_trend_over_time.csv")

# Create a line chart of the monthly sentiment trend
trend[order].plot(
    figsize=(9, 4),
    color=["#E74C3C", "#F39C12", "#2ECC71"]
)

# Add the chart title and axis labels
plt.title("Reviews over Time by Sentiment")
plt.ylabel("Count")
plt.xlabel("Month")

# Adjust the layout and save the figure
plt.tight_layout()
plt.savefig("outputs/fig_trend_over_time.png", dpi=150)

# Display the chart
plt.show()
```

### Summary

```python
if "word_count" not in df.columns:
    df["word_count"] = df["cleaned_text"].str.split().str.len()

g = df.groupby("platform")
summary = pd.DataFrame({
    "reviews":    g["review_id"].count(),
    "avg_rating": g["rating"].mean().round(2),
    "avg_words":  g["word_count"].mean().round(1),
    "%positive":  g["sentiment_label"].apply(lambda s: (s=="positive").mean()*100).round(1),
    "%neutral":   g["sentiment_label"].apply(lambda s: (s=="neutral").mean()*100).round(1),
    "%negative":  g["sentiment_label"].apply(lambda s: (s=="negative").mean()*100).round(1),
})
overall = pd.DataFrame({
    "reviews":[len(df)], "avg_rating":[round(df["rating"].mean(),2)],
    "avg_words":[round(df["word_count"].mean(),1)],
    "%positive":[round((df["sentiment_label"]=="positive").mean()*100,1)],
    "%neutral":[round((df["sentiment_label"]=="neutral").mean()*100,1)],
    "%negative":[round((df["sentiment_label"]=="negative").mean()*100,1)],
}, index=["ALL"])
summary = pd.concat([summary, overall])
summary.to_csv("outputs/table_summary_statistics.csv")
print(summary)
```
